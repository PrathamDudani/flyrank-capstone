## Visual sanity check: a handful of pages over time
Different pages clearly show different underlying trajectories (growth, decline, volatility, step-change around a refresh event) — this is the structure the modeling stage needs to pick up on.

```python
import matplotlib.pyplot as plt
sample_pages = df.page_id.drop_duplicates().sample(6, random_state=3)
fig, ax = plt.subplots(figsize=(9,4))
for p in sample_pages:
    g = df[df.page_id==p].sort_values('week')
    ax.plot(g.week, g.clicks, label=p, alpha=0.8)
ax.set_xlabel('week'); ax.set_ylabel('clicks'); ax.legend(fontsize=7); ax.set_title('Sample page trajectories')
plt.tight_layout(); plt.show()
```
